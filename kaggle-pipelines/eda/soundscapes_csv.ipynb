```python
import pandas as pd
import matplotlib.pyplot as plt
trs = pd.read_csv("/home/legionlinux/miniconda3/envs/torchenv/__INIT__/Kaggle/birdclef-2026/data/raw/train_soundscapes_labels.csv")
```

```python
trs
```

```python
counts = trs['primary_label'].value_counts(ascending=True)

print(counts)

fig, ax = plt.subplots(figsize=(10, 40))

counts.plot(kind='barh', ax=ax)

# Put count at the front/end of each bar
for i, v in enumerate(counts.values):
    ax.text(
        v,            # end of bar
        i,            # bar position
        f' {v}',      # count text
        va='center',
        ha='left'
    )

ax.set_xlabel('Count')
ax.set_ylabel('Primary Label')
ax.set_title('Distribution of Primary Labels')

plt.grid(True)
plt.show()
```
